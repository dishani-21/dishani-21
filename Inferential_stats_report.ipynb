```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv("Zingaro_Company.csv")
df.columns = df.columns.str.strip()

polished = df['Treated and Polished'].dropna()
unpolished = df['Unpolished'].dropna()

t_stat, p_value = ttest_ind(polished, unpolished, alternative='greater')

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Zingaro is justified in believing that unpolished stones may not be suitable for printing.")
    print("There is statistically significant evidence that unpolished stones are softer than treated and polished stones.")
else:
    print("Zingaro is not justified in claiming unpolished stones are unsuitable for printing.")
    print("There is no statistically significant evidence that unpolished stones are softer than treated and polished stones.")

```

```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv("Zingaro_Company.csv")
df.columns = df.columns.str.strip()

polished = df['Treated and Polished'].dropna()
unpolished = df['Unpolished'].dropna()

t_stat, p_value = ttest_ind(polished, unpolished, alternative='two-sided')

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("There is a statistically significant difference in mean hardness between treated and polished stones and unpolished stones.")
else:
    print("There is no statistically significant difference in mean hardness between the two groups.")
```

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

df = pd.read_excel("Dental Hardness data.xlsx")
df.columns = df.columns.str.strip()

for alloy in df['Alloy'].unique():
    print(f"\n🔍 ANALYSIS FOR ALLOY: {alloy}")

    df_alloy = df[df['Alloy'] == alloy]

    model = ols('Response ~ C(Method) + C(Temp) + C(Method):C(Temp)', data=df_alloy).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    print("\nANOVA Table:")
    print(anova_table)

    p_method = anova_table.loc['C(Method)', 'PR(>F)']
    p_temp = anova_table.loc['C(Temp)', 'PR(>F)']
    p_interaction = anova_table.loc['C(Method):C(Temp)', 'PR(>F)']

    if p_method < 0.05:
        print("Method has a significant effect on Response.")
    else:
        print("Method does NOT have a significant effect on Response.")

    if p_temp < 0.05:
        print("Temperature has a significant effect on Response.")
    else:
        print("Temperature does NOT have a significant effect on Response.")

    if p_interaction < 0.05:
        print("There is a significant interaction effect between Method and Temperature.")
    else:
        print("There is NO significant interaction between Method and Temperature.")
```

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

df = pd.read_excel("Dental Hardness data.xlsx")
df.columns = df.columns.str.strip()

for alloy in df['Alloy'].unique():
    print(f"One-Way ANOVA for {alloy} based on Method")

    df_alloy = df[df['Alloy'] == alloy]

    model = ols('Response ~ C(Method)', data=df_alloy).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)

    print("\nANOVA Table:")
    print(anova_table)

    p_value = anova_table['PR(>F)'].iloc[0]


    if p_value < 0.05:
        print("There is a significant difference in hardness between at least two Methods.")
    else:
        print("There is no significant difference in hardness between Methods.")

```
